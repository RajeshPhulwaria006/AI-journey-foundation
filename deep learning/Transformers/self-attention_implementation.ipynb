```python
import numpy as np

def to_embedding(text):
  tokens = text.lower().split()

  vocab = sorted(set(tokens))
  word_idx = {word: idx for idx, word in enumerate(vocab)}

  # embedding matrix
  embedding_dim = 5
  embedding_matrix = np.random.randn(len(vocab), embedding_dim)

  # convert sentence into embedding
  sentence_embeddings = []

  for token in tokens:
      idx = word_idx[token]
      sentence_embeddings.append(embedding_matrix[idx])

  sentence_embeddings = np.array(sentence_embeddings)
  return sentence_embeddings
```

```python
embedding_vector = to_embedding("I love machine learning")
embedding_vector
```

## Defining self-attention

```python
def softmax(x):
    exp = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp / np.sum(exp, axis=-1, keepdims=True)
```

```python
def initialize_weights(embedding_dim, d_k, d_v=None):
    if d_v is None:
        d_v = d_k

    w_Q = np.random.randn(embedding_dim, d_k)
    w_K = np.random.randn(embedding_dim, d_k)
    w_V = np.random.randn(embedding_dim, d_v)

    return w_Q, w_K, w_V
```

```python
def attention(embedding):
    # embedding: (seq_len, embedding_dim)
    embedding = np.asarray(embedding)
    w_Q, w_K, w_V = initialize_weights(embedding.shape[1], embedding.shape[1])

    # Generate Q, K, V
    Q = embedding @ w_Q
    K = embedding @ w_K
    V = embedding @ w_V

    # Compute Attention Scores
    seq_len = len(embedding)
    d_k = K.shape[1]
    scores = (Q @ K.T) / np.sqrt(d_k)

    # Softmax
    weights = softmax(scores)

    # Weighted Sum of Values
    output = weights @ V
    return output, weights, scores, Q, K, V
```

```python
output, weights, scores, Q, K, V = attention(embedding_vector)
```
