```python
import os
import gensim
import nltk
from gensim.utils import simple_preprocess
from nltk import sent_tokenize
```

```python
story = []
path = "/kaggle/input/datasets/khulasasndh/game-of-thrones-books/"
for filename in os.listdir(path):
    file = open(os.path.join(path, filename), encoding="cp1252")
    corpus = file.read()
    raw_tokens = sent_tokenize(corpus)
    for sentence in raw_tokens:
        story.append(simple_preprocess(sentence))
file.close()
```

```python
len(story)
```

```python
from gensim.models import Word2Vec

model = Word2Vec(
    window=10,
    min_count=2
)
```

```python
model.build_vocab(story)
```

```python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
model.wv.most_similar("daenerys")
```

```python
model.wv.doesnt_match(['jon', 'robb', 'bran', 'arya', 'sansa'])
```

```python
model.wv.get_normed_vectors().shape
```

```python
y = model.wv.index_to_key
len(y)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=3)
```

```python
X = pca.fit_transform(model.wv.get_normed_vectors())
```
