# Text Representation / Vectorization
> converting raw text into numerical vectors before it can be used for training or prediction.
- One Hot Encoding
- Bag of Words
- TF-IDF
- Word2Vec

```python
# one hot encoding
from sklearn.preprocessing import OneHotEncoder
import numpy as np

words = np.array([["I"], ["love"], ["NLP"], ["AI"]])

encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(words)

print(encoded)
```

```python
# Bag of Words
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "I love NLP",
    "I love AI"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
# TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "people watch campusx",
    "campusx watch campusx",
    "people write comment",
    "campusx write comment"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
# Word2Vec
from gensim.models import Word2Vec

# Sample corpus (already tokenized)
sentences = [
    ["i", "love", "machine", "learning"],
    ["machine", "learning", "is", "fun"],
    ["word2vec", "creates", "word", "embeddings"],
    ["deep", "learning", "uses", "embeddings"],
    ["i", "love", "deep", "learning"]
]

# Train Word2Vec model - CBoW
cBoW_model = Word2Vec(
    sentences=sentences,
    vector_size=100,   # Embedding dimension
    window=3,          # Context window size
    sg=0,              # 0 = CBOW
    epochs=10
)
```

```python
# skip-gram
skipGram_model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=1,          # Skip-Gram
    epochs=10
)
```
