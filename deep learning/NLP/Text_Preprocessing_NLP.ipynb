## 2. Remove HTML tags


```python
from bs4 import BeautifulSoup

text = """<h1>Welcome</h1>
<p>I love NLP</p>"""
print("Acutal text:\n", text)
```

```python
cleaned = BeautifulSoup(text, "html.parser").get_text()
print("Cleaned text:\n", cleaned)
```

## 3. Remove URLs

```python
import re
text = "Visit https://www.openai.com now!"
cleaned = re.sub(r'https?://\S+|www\.\S+', '', text)
print(cleaned)
```

## 4. Remove punctuations

```python
import string

# standard way
text = "Hello! How are you?"
trans = str.maketrans('', '', string.punctuation)
print(text.translate(trans))
```

```python
# manual way
def rem_punctuations(text):
  exclude = string.punctuation
  for char in exclude:
    text = text.replace(char, '')

  return text
```

```python
rem_punctuations(text)
```

## 5. Chat word treatment

```python
chat_dict = {
    "lol":"laughing out loud",
    "omg":"oh my god",
    "idk":"i do not know",
    "tbh":"to be honest",
    "btw":"by the way",
    "atp":"at this point"
}

text = "omg thats crazy tbh atp"
```

```python
# function to convert short form chat into full form
def conv_chat(text):
  words = text.split()
  expanded = [chat_dict.get(w, w) for w in words]
  return " ".join(expanded)
```

```python
conv_chat(text)
```

# 7. Stop Words Removal

```python
import nltk
from nltk.corpus import stopwords
nltk.download("stopwords")

text = "this is an example"

stop_words = set(stopwords.words("english"))
filtered = [w for w in text.split() if w not in stop_words]
print(" ".join(filtered))
```

## 9. Tokenization
 - sentence tokenization
 - word tokenization
 - character tokenization
 - subword tokenization

```python
from nltk.tokenize import word_tokenize, sent_tokenize
import nltk
nltk.download('punkt_tab')

# sentence tokenization
text = "Hello dear! Where are you today? nice to met you."

print("Manual way:", text.split('.'))   # this way doesnt recognise different sentences
print("Using function:", sent_tokenize(text))
```

```python
# word tokenization
text = "How are you?"

print("Manual:", text.split())    # this way doesnt recognize regular expression to split
print("Function:", word_tokenize(text))
```

```python
# character tokenization
text = "AI is good."

[i for i in text]
```

```python
# subword tokenization
from transformers import GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
print(tokenizer.tokenize("playing with you is awful"))
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

words = [
    "playing",
    "transformers",
    "electroencephalography",
    "internationalization",
    "unbelievably",
    "chatgptification",
    "RajeshGPTSuperModel"
]

for word in words:
    print(f"{word:30} -> {tokenizer.tokenize(word)}")
```

```python
# stemming tokenization
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

print(stemmer.stem("playing"))
print(stemmer.stem("studies"))
```

```python
# lemmatization
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("running", pos="v"))
print(lemmatizer.lemmatize("better", pos="a"))
```

### Note: - *The exact preprocessing pipeline depends on the NLP task. For example, modern transformer models like BERT and GPT generally require minimal preprocessing because they use sophisticated tokenizers (such as WordPiece or Byte Pair Encoding) and can effectively handle punctuation, casing (depending on the model), and subword units on their own.*
