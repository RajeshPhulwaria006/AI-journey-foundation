```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img
```

```python
keras.__version__
```

```python
img = load_img('/content/cat_img.jfif', target_size=(200, 200))
img
```

```python
type(img)
```

```python
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
)
```

```python
imgData = image.img_to_array(img)
type(imgData)
```

```python
img, imgData.shape
```

```python
input = imgData.reshape(1, 200, 200, 3)
```

```python
input.shape
```

```python
i = 0
for output in datagen.flow(input, batch_size=1, save_to_dir='/content/augmented'):
  i += 1
  if i == 10: break
```
